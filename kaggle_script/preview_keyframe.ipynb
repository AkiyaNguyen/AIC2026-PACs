# AIC2026 — preview one keyframe still

Open `VIDEO_ID` + `row_idx_in_video` (the search hit field `kf=` / `row_idx_in_video`, **0-based**) and display `{idx:06d}.webp`.

Example: `L22_V014` `kf=158` → `000158.webp` (not `000157.webp`). Map column `n` is 1-based (`n = row + 1`). `frame_idx` is the mp4 frame id (submission), not the filename.

1. Attach the keyframe dataset for that batch (`aic2026-keyframes-l**`). CPU is enough. No repo clone or `requirements.txt`.
2. Call `preview_keyframe(video_id, kf)` (`kf` = search `row_idx_in_video`, 0-based).


```python
# Dataset root = VIDEO_ID/{map.csv, *.webp, ...} (same as uploaded aic2026-keyframes-l**).
# Preview only needs the batch of VIDEO_ID (L22_V014 → attach L22). Extra batches are optional.

KEYFRAME_ROOTS = [
    "/kaggle/input/datasets/akiyanguyen/aic2026-keyframes-l21",
    "/kaggle/input/datasets/akiyanguyen/aic2026-keyframes-l22",
    "/kaggle/input/datasets/akiyanguyen/aic2026-keyframes-l23",
    "/kaggle/input/datasets/akiyanguyen/aic2026-keyframes-l24",
    "/kaggle/input/datasets/akiyanguyen/aic2026-keyframes-l25",
    "/kaggle/input/datasets/akiyanguyen/aic2026-keyframes-l26",
    "/kaggle/input/datasets/akiyanguyen/aic2026-keyframes-l27",
    "/kaggle/input/datasets/akiyanguyen/aic2026-keyframes-l28",
    "/kaggle/input/datasets/akiyanguyen/aic2026-keyframes-l29",
    "/kaggle/input/datasets/akiyanguyen/aic2026-keyframes-l30",
]
```

```python
import csv
from pathlib import Path

from IPython.display import Markdown, display
from PIL import Image


def preview_keyframe(video_id: str, kf: int) -> None:
    """Show `{kf:06d}.webp` for a search hit (`row_idx_in_video`, 0-based)."""
    folder = None
    for root in KEYFRAME_ROOTS:
        root_p = Path(root)
        if not root_p.is_dir():
            continue
        direct = root_p / video_id
        if (direct / "map.csv").is_file():
            folder = direct
            break
        for map_csv in root_p.rglob("map.csv"):
            if map_csv.parent.name == video_id:
                folder = map_csv.parent
                break
        if folder is not None:
            break
    if folder is None:
        raise FileNotFoundError(f"{video_id}: no map.csv under attached KEYFRAME_ROOTS")

    webp = folder / f"{kf:06d}.webp"
    if not webp.is_file():
        raise FileNotFoundError(webp)

    with (folder / "map.csv").open(encoding="utf-8", newline="") as f:
        rows = list(csv.DictReader(f))
    if kf < 0 or kf >= len(rows):
        raise IndexError(f"{video_id}: kf={kf} out of range 0..{len(rows) - 1}")
    meta = rows[kf]

    display(
        Markdown(
            f"**{video_id}** `row_idx_in_video={kf}` → `{webp.name}`  "
            f"(map `n`={meta['n']}, `pts_time`={meta['pts_time']}s, "
            f"`frame_idx`={meta['frame_idx']})"
        )
    )
    display(Image.open(webp))
    print("folder", folder)
    print("webp  ", webp)
    print("map row", meta)


preview_keyframe("L22_V014", 158)
```
